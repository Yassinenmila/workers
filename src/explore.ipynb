## **Visualisation Des Donnees**

### 1. Combien de client en rester ou abondoner


```python
df.hist(figsize=(12, 8), bins=30)
plt.tight_layout()
plt.show()
```

```python
sns.countplot(data=df, x="Churn")
plt.show()
```

### 2. churn selon le contrat

```python
sns.countplot(data=df, x="Contract", hue="Churn")

plt.xticks(rotation=45)
plt.show()
```

### 3.les charge mensuel par client 

```python
sns.histplot(data=df,x="MonthlyCharges",kde=True)

plt.show()
```

### 4. un affichage de relation entre monthly charges et chrun 

```python
sns.boxplot(data=df, x="Churn", y="MonthlyCharges")

plt.show()
```

```python
corr = df.select_dtypes(include='number').drop('SeniorCitizen',axis=1).corr()
sns.heatmap(corr, annot=True)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="tenure",
    y="MonthlyCharges",
    hue="Churn"
)
```

```python
sns.histplot(data=df, x="tenure", kde=True)

plt.title("Distribution de l'ancienneté")
plt.xlabel("Ancienneté")
plt.ylabel("Nombre de clients")
plt.show()
```

```python
corrr= df.corr()

plt.figure(figsize=(12,8))
sns.heatmap(corrr,annot=True,cmap='coolwarm',fmt='.2f')
plt.show()
```
