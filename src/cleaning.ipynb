```python
num_col= df.select_dtypes(include="number").columns

print(num_col)


for c in num_col:
    Q1=df[c].quantile(0.25)
    Q3=df[c].quantile(0.75)

    iqr = Q3 - Q1

    l = Q1 - 1.5 * iqr
    u = Q3 + 1.5 * iqr

    outliers = df[
        (df[c] < l) |
        (df[c] > u)
    ]

    print(f"{c}: {len(outliers)} outliers")
```
